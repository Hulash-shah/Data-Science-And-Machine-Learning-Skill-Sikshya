# Task 1

1. Parameters vs Hyperparameters

    Parameters are values learned automatically by the model during training.

    For Logistic Regression, the coefficients are parameters.

    Hyperparameters are settings that we choose before training. They control how the model learns.

Examples for Logistic Regression:

- C → controls regularization strength
- penalty → type of regularization
- solver → optimization algorithm

___

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split


df = pd.read_csv("Employee.csv")
```

```python
numeric_cols = ['Age', 'DailyRate', 'DistanceFromHome', 'Education',
       'EnvironmentSatisfaction', 'HourlyRate',
       'JobInvolvement', 'JobLevel', 'JobSatisfaction', 'MonthlyIncome',
       'MonthlyRate', 'NumCompaniesWorked', 'PercentSalaryHike',
       'PerformanceRating', 'RelationshipSatisfaction',
       'StockOptionLevel', 'TotalWorkingYears', 'TrainingTimesLastYear',
       'WorkLifeBalance', 'YearsAtCompany', 'YearsInCurrentRole',
       'YearsSinceLastPromotion', 'YearsWithCurrManager']

categorical_cols = [ 'BusinessTravel', 'Department', 'EducationField', 'Gender',
       'JobRole', 'MaritalStatus', 'OverTime']
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numeric_cols),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
    ]
)
```

```python
baseline_model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))
])

X = df.drop(columns = ['Over18','EmployeeCount',
       'EmployeeNumber','StandardHours','Attrition'])

y = df["Attrition"].map({
    "No": 0,
    "Yes": 1
})

X_train,X_test,y_train,y_test = train_test_split(X,y, test_size= 0.20 , random_state= 999)

baseline_model.fit(X_train, y_train)

baseline_pred = baseline_model.predict(X_test)
baseline_prob = baseline_model.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

print("Accuracy:", accuracy_score(y_test, baseline_pred))
print("Precision:", precision_score(y_test, baseline_pred))
print("Recall:", recall_score(y_test, baseline_pred))
print("F1 Score:", f1_score(y_test, baseline_pred))
print("ROC-AUC:", roc_auc_score(y_test, baseline_prob))
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    "classifier__C": [0.01, 0.1, 1, 10, 100],
    "classifier__penalty": ["l1", "l2"],
    "classifier__solver": ["liblinear"]
}
```

```python
grid_search = GridSearchCV(
    baseline_model,
    param_grid=param_grid,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters:", grid_search.best_params_)
print("Best CV Score:", grid_search.best_score_)
```

```python
grid_pred = grid_search.predict(X_test)
grid_prob = grid_search.predict_proba(X_test)[:, 1]
```

```python
print("GridSearch F1:", f1_score(y_test, grid_pred))
print("GridSearch ROC-AUC:", roc_auc_score(y_test, grid_prob))
```

___
___

```python
from sklearn.model_selection import RandomizedSearchCV

param_dist = {
    "classifier__C": [0.001, 0.01, 0.1, 1, 10, 100, 1000],
    "classifier__penalty": ["l1", "l2"],
    "classifier__solver": ["liblinear"]
}
```

```python
random_search = RandomizedSearchCV(
    baseline_model,
    param_distributions=param_dist,
    n_iter=10,
    cv=5,
    scoring="f1",
    random_state=42,
    n_jobs=-1
)

random_search.fit(X_train, y_train)
```

```python
print("Best Parameters:", random_search.best_params_)
print("Best CV Score:", random_search.best_score_)
```

```python
random_pred = random_search.predict(X_test)
random_prob = random_search.predict_proba(X_test)[:, 1]

print("RandomizedSearch F1:", f1_score(y_test, random_pred))
print("RandomizedSearch ROC-AUC:", roc_auc_score(y_test, random_prob))
```

___
___

```python
results = pd.DataFrame({
    "Model": ["Baseline", "GridSearchCV", "RandomizedSearchCV"],
    "F1 Score": [
        f1_score(y_test, baseline_pred),
        f1_score(y_test, grid_pred),
        f1_score(y_test, random_pred)
    ],
    "ROC-AUC": [
        roc_auc_score(y_test, baseline_prob),
        roc_auc_score(y_test, grid_prob),
        roc_auc_score(y_test, random_prob)
    ]
})

print(results)
```

The baseline Logistic Regression achieved an F1-score of 0.5195 and a ROC-AUC of 0.7791. After applying GridSearchCV, the F1-score remained the same at 0.5195, while the ROC-AUC decreased slightly to 0.7790. RandomizedSearchCV produced an F1-score of 0.5063 and a ROC-AUC of 0.7764, which were lower than the baseline. Therefore, hyperparameter tuning did not improve the model's performance in this experiment. This shows that tuning does not always lead to better test-set performance, especially when the baseline model is already reasonably well configured. GridSearchCV also requires more computation because it evaluates all specified combinations, whereas RandomizedSearchCV evaluates only a selected number of combinations.

___
GridSearchCV is more exhaustive but can be computationally expensive, while RandomizedSearchCV is generally faster because it tests only a limited number of randomly selected combinations. Hyperparameter tuning can improve model performance, but the improvement should be compared with the additional computational cost. For a small search space, GridSearchCV is practical, while RandomizedSearchCV can be more efficient for a large search space.
