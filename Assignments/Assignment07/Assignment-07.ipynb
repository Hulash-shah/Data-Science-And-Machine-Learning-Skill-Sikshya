```python
def count_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return ((df[column] < lower) | (df[column] > upper)).sum()

outlier_counts = {col : count_outliers(house_pricing_df, col) for col in numerical_types}
outlier_summary = pd.Series(outlier_counts).sort_values(ascending = False)
print(outlier_summary)
```

**Findings:** Running the IQR method across all numerical columns revealed
that columns like EnclosedPorch, BsmtFinSF2, and ScreenPorch show the highest
raw outlier counts — but this is largely because these features are absent
(value = 0) for most houses, making any non-zero value appear as a statistical
outlier. OverallCond, while flagged, is an ordinal rating rather than a
continuous measurement, so IQR-based outlier detection doesn't meaningfully
apply to it. The most genuinely informative outliers appear in continuous
size-related measurements like LotFrontage, TotalBsmtSF, and GrLivArea,
which are examined further below.

```python
zero_pct = (house_pricing_df[numerical_types] == 0).mean().sort_values(ascending=False) * 100
print(zero_pct.head(10))
```

**Findings:** Several columns (EnclosedPorch, BsmtFinSF2, ScreenPorch, MasVnrArea,
PoolArea) showed high "outlier" counts via the IQR method, but this is a
consequence of these features being zero-inflated — most houses simply lack
these features (e.g., no pool), making any non-zero value appear statistically
extreme. These zero values represent genuine, meaningful data and should NOT
be removed. Instead, more targeted outlier investigation was applied to
continuous measurements like GrLivArea, where a small number of points
showed a genuine anomaly (unusually large living area paired with low

### **Problem 3: Handling Missing Data and Imputation**  
1. Identify columns with **missing values**.
2. Impute the missing values:
   - For **numerical columns**, replace missing values with the **median**.
   - For **categorical columns**, replace missing values with the **most frequent** value.
3. Justify your choice of imputation methods and check if any missing data remains.

```python
house_pricing_df.isna().sum()
```

```python
if 'SalePrice' in numerical_types:
    numerical_types.remove('SalePrice')

# Impute numerical columns with median
for col in numerical_types:
    if house_pricing_df[col].isna().sum() > 0:
        median_val = house_pricing_df[col].median()
        house_pricing_df[col] = house_pricing_df[col].fillna(median_val)

# Impute categorical columns with mode (most frequent value)
for col in categorical_types:
    if house_pricing_df[col].isna().sum() > 0:
        mode_val = house_pricing_df[col].mode()[0]
        house_pricing_df[col] = house_pricing_df[col].fillna(mode_val)
```

```python
missing_remain = house_pricing_df.isna().sum().sum()
missing_remain
```

**Justification for imputation methods:**

**Median for numerical columns:** The median was chosen over the mean because
several numerical columns (e.g., LotArea, GrLivArea, LotFrontage) are
right-skewed, as identified in the earlier outlier analysis. The mean is
sensitive to extreme values and would be pulled upward by outliers, producing
an imputed value that doesn't represent a "typical" house. The median is
robust to skew and outliers, making it a safer default for filling missing
numerical values without distorting the overall distribution.

**Mode for categorical columns:** For categorical features, there is no
meaningful "average" — the most frequent category is the most reasonable
assumption for a missing value, since it represents the most common/likely
value observed in the dataset.

**Caveat:** Several columns in this dataset (e.g., PoolQC, Alley, Fence,
FireplaceQu) use NaN specifically to indicate the *absence* of a feature
(e.g., no pool), rather than truly missing/unknown data, as documented in
data_description.txt. Applying mode imputation to these columns technically
fulfills the assignment's instructions, but conceptually overwrites
meaningful "feature does not exist" information with an artificial category.
In a real-world project, these columns would typically be filled with
"None" instead — this distinction is noted here for transparency, though
the assignment's specified method (mode imputation) was applied as instructed.

After imputation, verification confirmed 0 remaining missing values across
the dataset.
