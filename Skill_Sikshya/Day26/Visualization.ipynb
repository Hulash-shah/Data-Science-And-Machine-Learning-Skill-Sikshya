```python
import pandas as pd

titanic_df = pd.read_csv(
    r"C:\Users\shahh\OneDrive\Desktop\Data Science And Machine Learning\Skill_Sikshya\Day22\titanic_data.csv",
    delimiter = ",",
    header = 0
    )

titanic_df
```

Types of graph in pandas visualization

kind=	Chart
"line"	Line plot
"bar" / "barh"	Vertical/horizontal bar
"hist"	Histogram
"box"	Box plot
"kde" / "density"	Density curve
"area"	Area plot
"pie"	Pie chart
"scatter"	Scatter plot
"hexbin"	Hexbin density plot


1. Bar Graph

```python

titanic_df.value_counts(subset = [ "Sex", "Embarked"]).plot(
    kind = "bar",
    title = "Passenger Count by Embarkation Port",
    figsize = (5,5),
    xlabel = "Port of Embarkation( S, C, Q )",
    ylabel = "No. of Passengers",
    grid = False,
    yticks = range(0, 700, 50),
    cmap = "viridis",
    # color = ["red", "green", "blue"]
)
```

```python
pd.crosstab(
    titanic_df.Embarked,
    titanic_df.Sex,
).plot(
    kind = "bar",
    title = "Passenger Count by Embarkation Port",
    figsize = (5,5),
    xlabel = "Ports of Embarkation( S, C, Q )",
    ylabel = "No. of Passengers",
    grid = False,
    yticks = range(0, 700, 50),
    cmap = "viridis",
    # color = ["red", "green", "blue"]
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.countplot(x="Embarked", hue="Sex", data=titanic_df)
plt.title("Passenger Count by Embarkation Port and Gender")
plt.xlabel("Port of Embarkation")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

ax = sns.countplot(x="Embarked", hue="Sex", data=titanic_df)
plt.title("Passenger Count by Embarkation Port and Gender")
plt.xlabel("Port of Embarkation")
plt.ylabel("Number of Passengers")

# Add count labels on top of each bar
for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

```python
ax = pd.crosstab(titanic_df["Embarked"], titanic_df["Sex"]).plot(kind="bar")
for container in ax.containers:
    ax.bar_label(container)
plt.show()
```

```python
import matplotlib.pyplot as plt

ax = titanic_df.value_counts(subset=["Embarked", "Sex"]).unstack().plot(
    kind="bar",
    title="Passenger Count by Embarkation Port",
    figsize=(6, 5),
    xlabel="Port of Embarkation (S, C, Q)",
    ylabel="No. of Passengers",
    grid=False,
    yticks=range(0, 500, 50),
    colormap="viridis",
)

for container in ax.containers:
    ax.bar_label(container)

plt.legend(title="Sex")
plt.tight_layout()
plt.show()
```

2. Pie Chart

```python
group_by_embarked = titanic_df.groupby(by = "Embarked")
group_by_embarked
```

```python
group_by_embarked.count().plot(
    kind = "pie",
    y = "Survived",
    title = "Count of Passengers according to Pclass",
    autopct = "%1.1f%%"
)
```

```python
crosstab = pd.crosstab(
    titanic_df.Survived,
    titanic_df.Sex,
)
crosstab
```

```python
crosstab.T.plot(
    kind = "pie",
    subplots = True,
    legend = False,
    autopct = "%1.1f%%"
)
```

```python
crosstab.T.plot(
    kind="pie",
    subplots=True,
    figsize=(10, 5),
    autopct="%1.1f%%",
    legend=False,
)
plt.suptitle("Passenger Class Distribution: Did Not Survive vs. Survived")
plt.tight_layout()
plt.show()
```

3. Histogram

```python
titanic_df["Age"].dropna().plot(
    kind="hist",
    bins=20,
    color="blue",
    edgecolor="black",
    title="Distribution of Passenger Age",
    xlabel="Age",
    ylabel="Number of Passengers"
)
```

```python
titanic_df.Age.plot(
    kind = "kde",
    grid = True,
    ind = range(0, 81),
    color = "blue",
)
```

```python
titanic_df[["Age", "Survived"]].plot(
    kind = "hist",
    by = "Survived",
    bins = 20,
    grid = False,
   
    color = "blue",
)
```

```python
titanic_df[["Age", "Fare"]].plot(
    kind = "kde",
    figsize = (12,3),
    ind = range(0, 200)
)
```
