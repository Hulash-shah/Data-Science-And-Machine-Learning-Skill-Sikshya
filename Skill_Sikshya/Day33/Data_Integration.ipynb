Merging

```python
import pandas as pd

df1 = pd.DataFrame(
    {"Key" : ["AB", "CD", "EF","AC"], "name" : ["Ram", "Shyam", "Hari", "Krishna"], "age" : [24,25,23,24]}
)

df2 = pd.DataFrame(
    {"Key" : ["AB","BD","CB","FE"], "principle" : [400,400,500,600], "interest" : [6.4, 5.6,4.5,8.9]}
)
```

```python
df1, df2
```

```python
pd.merge(df1,df2, how = "right", on = "Key")
```
