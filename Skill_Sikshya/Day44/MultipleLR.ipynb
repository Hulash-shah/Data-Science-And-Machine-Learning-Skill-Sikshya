```python
import pandas as pd
import numpy as np


data_df = pd.read_csv(r"C:\Users\shahh\OneDrive\Desktop\Data Science And Machine Learning\Skill_Sikshya\Day43\Advertising.csv", 
                      index_col = 0)
data_df.head(5)
```

```python
X = data_df.drop("Sales", axis = 1).to_numpy()
y = data_df[["Sales"]].to_numpy()
```

```python
X.shape
```

```python
y.shape
```

```python
X = np.concat(
    (np.ones((X.shape[0],1)), X),
    axis = 1
)
X.shape
```

```python
betas = np.linalg.inv(np.dot(X.T, X)).dot(X.T).dot(y)
betas
```

```python
betas = betas.flatten()
print(betas)
```

```python
print(f"y = {betas[0]:.3f} + {betas[1]:.3f} x1 + {betas[2]:.3f} x2 + {betas[3]:.3f} x3")
```

```python
from sklearn.model_selection import train_test_split 
```

```python

X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.20, random_state=987)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lin_reg = LinearRegression().fit(X_train,y_train)
```

```python
y_pred=lin_reg.predict(X_test)
```

```python
from sklearn.metrics import max_error, mean_absolute_error, mean_squared_error, r2_score
print(f"R2:{r2_score(y_test,y_pred)}")
print(f"Max Error:{max_error(y_test,y_pred)}")
print(f"Mean Asoulte Error:{mean_absolute_error(y_test,y_pred)}")
print(f"Mean squared Error :{mean_squared_error(y_test,y_pred)}")
```
