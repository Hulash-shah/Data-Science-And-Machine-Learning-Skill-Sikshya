Machine Learning
Machine Learning (ML) is a subfield of Artificial Intelligence where systems learn patterns from data rather than being explicitly programmed with fixed rules. Instead of writing step-by-step instructions for every scenario, you feed a model examples, and it learns to generalize from them — improving its predictions or decisions as it's exposed to more data.

The core idea: a traditional program takes rules + data → produces output. Machine Learning flips this: it takes data + output (examples) → produces the rules (a trained model) — which can then be applied to new, unseen data.

Machine Learning splits into three major approaches, based on how the model learns from data:

1.supervised Learning —> the model learns from labeled examples, where each input already has a known correct output. You show it thousands of "house features → price" or "email text → spam/not spam" pairs, and it learns the mapping. Once trained, it predicts labels for new, unseen inputs. This is the most common and easiest-to-evaluate approach, since you always have ground truth to check against. Algorithms: Linear/Logistic Regression, Decision Trees, Random Forests, SVMs, Neural Networks.

2.Unsupervised Learning —> there are no labels at all. The model is given raw data and asked to find structure on its own — grouping similar items together (clustering), reducing complexity while preserving important patterns (dimensionality reduction), or detecting unusual data points (anomaly detection). Since there's no "correct answer" to check against, evaluation is trickier and often more subjective. Algorithms: K-Means, Hierarchical Clustering, PCA, DBSCAN.

3.Reinforcement Learning —>an agent learns by interacting with an environment, taking actions, and receiving rewards or penalties as feedback. Instead of learning from a fixed dataset, it learns through trial and error over time, gradually discovering which actions maximize long-term reward. This is the approach behind game-playing AI (like AlphaGo), robotics, and increasingly, fine-tuning modern LLMs (RLHF). Algorithms: Q-Learning, Deep Q-Networks (DQN), Policy Gradients.

A fourth hybrid worth knowing: Semi-supervised learning sits between the first two — using a small amount of labeled data alongside a much larger pool of unlabeled data. This is increasingly common in practice, since labeling data is expensive but collecting raw data is cheap.
