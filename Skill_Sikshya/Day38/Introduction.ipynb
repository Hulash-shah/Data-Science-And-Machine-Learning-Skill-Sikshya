Linear Regression
Linear Regression is the simplest and most foundational algorithm in ML — it tries to draw the best straight line through your data points.

The core idea: find the straight line y = mx + b that best fits the scattered data points — minimizing the total distance between the line and every actual point.

The math, in plain terms:

y = the value you're predicting (e.g., house price)
x = your input feature (e.g., house size)
m (slope) = how much y changes for each unit increase in x
b (intercept) = the baseline value of y when x is 0

How the "best" line is found — Ordinary Least Squares (OLS):
The algorithm looks at the vertical distance between each actual data point and the line (called a "residual"), squares each of those distances (to penalize larger errors more and avoid positive/negative errors canceling out), and finds the exact line that makes the sum of all those squared distances as small as possible.

Assumptions Of Linear Regression

1. Linearity

What it means: The relationship between each independent variable (X) and the dependent variable (y) is actually linear — not curved, exponential, or some other shape.

Why it matters: If the true relationship is curved but you fit a straight line, the model will systematically over/under-predict in different regions — it's not just "a bit off," it's structurally wrong.

2. Independence of errors

What it means: The residual (error) for one observation shouldn't be correlated with the residual of another. This is especially important in time-series data, where today's error often depends on yesterday's.

Why it matters: Violating this (called autocorrelation) makes your standard errors unreliable — you might think your model is more confident than it actually is.

3. Homoscedasticity (equal variance of errors)

What it means: The spread (variance) of residuals should stay roughly constant across all levels of the predicted value — not fan out or shrink as predictions get larger.

Why it matters: If errors get bigger as predictions get bigger (a common pattern — e.g., predicting prices where expensive items have more absolute error), your model's confidence intervals become unreliable

4. Normality of residuals

What it means: The residuals (errors), not the raw data itself, should be approximately normally distributed.

Why it matters: This matters mainly for the statistical inference side (confidence intervals, p-values, hypothesis tests on coefficients) — not so much for raw prediction accuracy. With large sample sizes, this assumption matters less (Central Limit Theorem helps).

5. No (or low) multicollinearity

What it means: Independent variables shouldn't be too highly correlated with each other. If two features basically carry the same information, the model can't reliably tell which one is actually driving the outcome.

Why it matters: Coefficients become unstable and hard to interpret — small changes in data can wildly swing a coefficient's value or even its sign.

6. No significant outliers / high-leverage points

What it means: Extreme values shouldn't be disproportionately dragging the fitted line in their direction. This isn't always listed as a formal "assumption," but Linear Regression's squared-error cost function is very sensitive to outliers, so it's practically essential to check.



"""
PARAMETER ESTIMATION TECHNIQUES
================================

1. Ordinary Least Squares (OLS)
   - Minimizes sum of squared residuals: sum((y - y_pred)^2)
   - Closed-form solution: beta = (X^T X)^-1 X^T y
   - Used in: Linear Regression
   - sklearn: LinearRegression()

2. Maximum Likelihood Estimation (MLE)
   - Finds parameters that make observed data most probable
   - theta_hat = argmax P(data | theta)
   - Used in: Logistic Regression, Naive Bayes
   - sklearn: LogisticRegression()

3. Gradient Descent (+ variants)
   - Iteratively updates params: theta = theta - lr * gradient(cost)
   - Batch GD -> full dataset per step (slow, accurate)
   - SGD -> one sample per step (fast, noisy)
   - Mini-batch GD -> small batches (standard in deep learning)
   - Used in: Neural Networks, large-scale Logistic/Linear Regression
   - sklearn: SGDRegressor(), SGDClassifier()

4. Method of Moments
   - Matches sample statistics (mean, variance) to theoretical ones
   - Used in: classical statistics, distribution fitting

5. Bayesian Estimation (MAP)
   - MLE + a prior belief about parameters
   - theta_hat = argmax P(data|theta) * P(theta)
   - Used in: Bayesian Linear Regression, Bayesian Neural Nets




6. Regularized Estimation
   - Ridge (L2): cost + lambda * sum(beta^2)      -> shrinks coefficients
   - Lasso (L1): cost + lambda * sum(|beta|)      -> can zero out coefficients (feature selection)
   - sklearn: Ridge(alpha=1.0), Lasso(alpha=1.0)

SUMMARY
-------
OLS              -> exact, Linear Regression
MLE              -> probability-based, Logistic Regression
Gradient Descent -> iterative, Neural Networks / big data
Method of Moments-> classical stats
Bayesian/MAP     -> MLE + prior knowledge
Ridge/Lasso      -> OLS + penalty, handles overfitting/multicollinearity
