```python
one_hot_encoder = OneHotEncoder(sparse_output = False)
ordinal_encoder = OrdinalEncoder(dtype = int)
```

```python
age_embarked_encoded = one_hot_encoder.fit_transform(titanic_df[["Sex", "Embarked"]])
pclass_encoded = ordinal_encoder.fit_transform(titanic_df[["Pclass"]])
```

```python
final_df.Pclass = pclass_encoded
```

```python
final_df = pd.get_dummies(final_df, columns = ["Sex", "Embarked"], dtype = int)
```
