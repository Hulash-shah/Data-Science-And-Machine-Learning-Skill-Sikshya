1. Min - Max Normalization

```python
min_fare = titanic_df.Fare.min()
max_fare = titanic_df.Fare.max()
```

```python
norm_fare = (titanic_df.Fare - min_fare) / (max_fare - min_fare)
```

```python
norm_fare.describe()
```

```python
norm_fare.plot(
    kind = "kde"
)
```
