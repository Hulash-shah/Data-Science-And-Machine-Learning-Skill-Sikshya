Titanic Dataset – Age & Fare Analysis

• The dataset contains 891 passengers.
• Age has 714 valid values and 177 missing values.
• The average passenger age is 29.7 years.
• The median passenger age is 28 years.
• Passenger ages range from 0.42 to 80 years.
• The average fare is 32.20.
• The median fare is 14.45.
• Fare values range from 0 to 512.33.
• The difference between mean and median fare shows that Fare is right-skewed.
• The maximum fare of 512.33 indicates the presence of outliers.
• 75% of passengers paid a fare of 31 or less.
• Age and Fare show considerable variation among passengers.
• Missing Age values should be handled before further analysis or machine learning.

```python
display = pd.crosstab(
    index = titanic_df.Sex,
    columns = titanic_df.Survived,
    margins = True,
    margins_name = "Total",
    colnames = ["Rate of Survival"],
    rownames = ["Gender"],
    aggfunc = "mean",
    values = titanic_df.Fare
)
print(display)
```

```python
dis = pd.crosstab(
    titanic_df.Sex,
    [titanic_df.Survived, titanic_df.Pclass],
    margins = False
    
)
print(dis)
```
