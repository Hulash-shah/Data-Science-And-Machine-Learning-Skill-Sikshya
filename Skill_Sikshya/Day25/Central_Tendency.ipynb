```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
x = [10, 20, 30, 40, 50]
```

```python
_mean = np.mean(x)
print(f"Mean of x is :", _mean)
```

```python
_median = np.median(x)
print(f"Median of x is :", _median)
```

Mode


```python
import statistics

_mode = statistics.mode(x)
print(f"Mode of x is:", _mode)
```
