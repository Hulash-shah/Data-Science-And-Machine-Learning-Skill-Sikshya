Removing Null Values

```python
titanic_df.isna().sum()
```

```python
titanic_df.dropna(subset = "Embarked", axis = 0, ignore_index = True, inplace = True)
titanic_df.shape
```

```python
# Null value as Feature
titanic_df.Cabin = titanic_df.Cabin.notna().astype(int)
```

```python
titanic_df.isna().sum()
```

# Fill null Values

```python
titanic_df.Age.plot(
    kind = "hist",
    bins =  20,
)
```

```python

titanic_df["Age"] = titanic_df["Age"].fillna(titanic_df["Age"].median())
titanic_df["Age"]
```

```python
titanic_df.Age.plot(
    kind = "hist",
    bins = 20,
)
```

```python
titanic_df.plot(kind="box", column="Age", by="Pclass", figsize=(8,5), grid=True)
```

```python
# Inputation of Age with respect to Gender
titanic_df["Age"] = titanic_df.groupby("Sex")["Age"].transform(lambda x: x.fillna(x.median()))
titanic_df.head()
```
