```python
import pandas as pd
import seaborn as sns

titanic_df = pd.read_csv(
    r"C:\Users\shahh\OneDrive\Desktop\Data Science And Machine Learning\Skill_Sikshya\Day22\titanic_data.csv",
    delimiter = ",",
    header = 0
    )

titanic_df
```

6. Box Plot
   

```python
titanic_df[["Age", "Fare"]].describe()
```

```python
titanic_df.plot(
    kind = "box",
    column = "Age",
    vert = False,
    figsize = (10, 3),
    xticks = range(0, 84, 4),
    grid = True
)
```

```python
titanic_df.plot(
    kind="box",
    column="Age",
    vert=False,
    figsize=(10, 3),
    xticks=range(0, 84, 4),
    grid=True,
    title="Distribution of Passenger Age",
    color={
        "whiskers": "gray",
        "medians": "black",
        "caps": "gray",
        "boxes": "#4C72B0"
    },
    patch_artist=True,   # fills the box with color instead of leaving it hollow
    sym="r+",             # outliers shown as red + marks instead of plain circles
)
```

```python
titanic_df.plot(
    kind="box",
    column="Fare",
    vert=False,
    by = "Sex",
    grid=True,
    figsize=(12, 3),
    xticks=range(0, 600, 50),
    title="Distribution of Passenger Fare (Log Scale)",
    color={
        "whiskers": "gray",
        "medians": "black",
        "caps": "gray",
        "boxes": "#4C72B0"
    },
    patch_artist=True,
    sym="r+"
)
```

7. Heat Map

```python
correlation = titanic_df.corr(numeric_only = True)
correlation
```

```python
import numpy as np
mask = np.triu(np.ones_like(correlation, dtype = bool))
mask
```

```python
sns.heatmap(data = correlation, cmap = "coolwarm", fmt = ".2f", annot = True, )
```

```python
sns.heatmap(
    data = correlation,
    cmap = "coolwarm",
    fmt = ".2f",
    annot = True,
    mask = mask )
```
