Chi-Square Test
The Chi-Square Test (χ²) is used to determine whether there is a significant relationship between categorical variables by comparing observed and expected frequencies.

Formula:

$$ \chi^2 = \sum \frac{(O-E)^2}{E} $$

Where:

O = Observed frequency
E = Expected frequency
χ² = Chi-square statistic

A small p-value (typically < 0.05) indicates a statistically significant relationship.


```python
import scipy.stats as stats
```

```python
contengency_table = pd.crosstab(titanic_df.Pclass, titanic_df.Survived)
contengency_table
```

```python
chi2_score, pvalue, dof, excepted_frquency = stats.chi2_contingency(contingency_table)
```

```python
print(f"Chi-Squared test score = {round(chi2_score, 3)}")
print(f"p-value = {round(pvalue, 5)}")
pvalue < 0.05
```
